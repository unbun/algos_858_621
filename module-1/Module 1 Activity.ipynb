# (a) Lists
- Define a "list" as comprehensively and accurately as you would find in a dictionary. Provide a clear and concise explanation, covering its essential meaning, characteristics, and any relevant context.
- Create a list in Python using the first observation, all features and label from the Iris data set
- Show the output with a print statement
- Add the second observation of the Iris data set to your list
- Show the output of your list with a print statement

A list is a mutable sequence of elements, ordered and accesible by index. Lists can contain elements of any types, can duplicate items, and supports a wide range of operations, such as appending, inserting, removing, sorting, etc. In Python, a list is written as a comma-seperated set of values between square brackets, and has a starting index of 0.

```python
import pandas as pd # for data management
from pprint import pprint # for pretty-printing dictionaries

from sklearn.datasets import load_iris
iris = load_iris()
iris_df = pd.DataFrame(
    data=iris.data,
    columns=iris.feature_names
)
iris_df['target'] = iris.target
```

```python
def obs(row):
  '''Helper method to get a single observation (i.e. row of all features and label) in a formatted way'''
  return row[iris.feature_names].astype(float).tolist() + [int(row['target'])]

# Create list and add 1st observation
l_observations = obs(iris_df.iloc[0])
print("1st Observation: ", l_observations)
  
# Add the second observation. Use extend so the list doesn't become nested
l_observations.extend(obs(iris_df.iloc[1]))
print("1st & 2nd Observation: ", l_observations)
```

# (b) Dictionary
- Define a "dictionary" as comprehensively and accurately as you would find in a dictionary. Provide a clear and concise explanation, covering its essential meaning, characteristics, and any relevant context.
- Create a dictionary in Python using the classes of Setosa, Versicolor and Virginica, allow elements (2 observations per class) to be added
- Show the output with a print statement
- Add a third observation of the Iris data set for each class
- Show the output of your with a print statement

A dictionary is a mutable unordered collection of key-value pairs, where each key maps to exactly one value. Keys are unique and immutable, but values can be any type (including lists or other dictionaries). In Python, they are written with curly braces where the key and value is seperated by a colon, and each pair is seperated by a comma.

```python
# Add data to the df for species name
name_map = {i: str(name) for i, name in enumerate(iris.target_names)}
iris_df['species'] = iris_df['target'].map(name_map)
cols = list(iris.feature_names) + ['target']

# Create a dictionary. For each species, get the 1st 2 dataframe elements in that species' group
# By making the values in the dictionary a list, we allow elements to be added. And now
# we make each observation a standalone element in the list, so we do want nested lists.
d_observations = {
  species: [obs(r) for _, r in group.head(2).iterrows()]
  for species, group in iris_df.groupby('species')
}   
print("2 Observations per class:")
pprint(d_observations, sort_dicts=False, width=60)
print()

# Add 3rd Observations per species
for species, group in iris_df.groupby('species'):
  d_observations[species].append(obs(group.iloc[2]))
print("3 Observations per class:")
pprint(d_observations, sort_dicts=False, width=60)
```

# (d) Set
- Define a "set"" as comprehensively and accurately as you would find in a dictionary. Provide a clear and concise explanation, covering its essential meaning, characteristics, and any relevant context.
- Create a set (1st set) in Python using the first observation of the Setosa class allowing elements to be added
- Show the output with a print statement
- Create another set (2nd set) by adding the second observation of the Setosa class 
- Show the output of the following operations with the 1st set and 2nd set with a print statement
    * Difference
    * Intersection
    * Union

A set is a collection of unique, unordered elements. Sets operations include Union, Intersection, and Difference. In python, a set is also a data type, and can be declared with set() or set(list). 

```python
# Declare and print sets with observations added
s1 = set(d_observations['setosa'][0])
print("Set 1: ", s1)

s2 = set(d_observations['setosa'][1])
print("Set 2: ", s2)

# Perform set operations and print
print("S1 \\ S2: ", s1.difference(s2))
print("S1 ∩ S2: ", s1.intersection(s2))
print("S1 ∪ S2: ", s1.union(s2))
```
